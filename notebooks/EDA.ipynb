## import librarise and setup

```python
#EDA 
import numpy as np
import pandas as pd
# time stamp for last update to maintain the records of updates
import time as t
print('Last Update:',t.asctime())



#filter warnings (to ignore warning from third party librariese)
import warnings as w
w.filterwarnings('ignore')

#set random seed (Setup random seed to ingone the redundancy in random data)
np.random.seed(42)
```

## Dataset

### Generate Data Using numpy

```python
names =np.char.add('std',np.arange(1,101).astype('str'))
names[:10]
```

```python
ages=np.random.randint(18,40,100)
ages[:10]
```

```python
gender=np.random.choice(['Male','Female'],100)
gender
```

```python
ml_100=np.random.randint(20,100,100)
dl_100=np.random.randint(10,100,100)
nlp_100=np.random.randint(40,100,100)
eda_60=np.random.randint(50,61,100)
```

```python
p1 = np.random.randint(1000, 99999, 100).astype(str)
p2 = np.random.randint(1000000, 9999999, 100).astype(str)
p3 = np.random.randint(0, 10, 100).astype(str)

p1 = np.char.add(p1, '-')
p2 = np.char.add(p1, p2)
p3 = np.char.add('-', p3)

cnic = np.char.add(p2, p3)

cnic[:1]
```

### introduce null values in data

```python
cnic[np.random.randint(0,100,5)]=None
cnic
```

# Convert it into Dictionary

```python
data={
    "Name":names,
    'Age':ages,
    'Gender':gender,
    'CNIC':cnic,
    'Machine Learning':ml_100,
    'Deep Learning':dl_100,
    'Nlp':nlp_100,
    'EDA':eda_60
}
```

## convert into data frame

```python
df=pd.DataFrame(data)
df.head()
```

```python
df.to_csv('student_records.csv')
```

## Dataset

### Generate Data Using numpy

```python
names =np.char.add('std',np.arange(1,101).astype('str'))
names[:10]
```

```python
ages=np.random.randint(18,40,100)
ages[:10]
```

```python
gender=np.random.choice(['Male','Female'],100)
gender
```

```python
ml_100=np.random.randint(20,100,100)
dl_100=np.random.randint(10,100,100)
nlp_100=np.random.randint(40,100,100)
eda_60=np.random.randint(50,61,100)
```

```python
p1 = np.random.randint(1000, 99999, 100).astype(str)
p2 = np.random.randint(1000000, 9999999, 100).astype(str)
p3 = np.random.randint(0, 10, 100).astype(str)

p1 = np.char.add(p1, '-')
p2 = np.char.add(p1, p2)
p3 = np.char.add('-', p3)

cnic = np.char.add(p2, p3)

cnic[:1]
```

### introduce null values in data

```python
cnic[np.random.randint(0,100,5)]=None
cnic
```

## Convert it into Dictionary

```python
data={
    "Name":names,
    'Age':ages,
    'Gender':gender,
    'CNIC':cnic,
    'Machine Learning':ml_100,
    'Deep Learning':dl_100,
    'Nlp':nlp_100,
    'EDA':eda_60
}
```

## convert into data frame

```python
df=pd.DataFrame(data)
df.head()
```

```python
df.to_csv('student_records.csv',index=False)
```

### Feature Engineering

```python
df['Obtained Marks']=df['Machine Learning']+df['Deep Learning']+df['Nlp']+df['EDA']
df.head(1)
```

```python
df['Percentage']=(df['Obtained Marks']*100/100).round(2)
df.head(1)
```

```python
df['Average']=(df['Obtained Marks']/4).round(2)
df.head(1)
```

```python
def check_status(x):
    if(x>50):
        return 'Passed'
    else:
        return 'Failed'
```

```python
df['Status']=df['Percentage'].apply(check_status)
df.head(1)
```

```python
df['Deep Learning'].max()
```

```python
# min marks w.r.t gender in subject
df.groupby('Gender')['Deep Learning'].min()
```

# Product Sale Ratio

```python
product=['tshirt','casual-shirt','jeans','dress-shirt']
sales=[23,64,12,19]
```

```python
plt.bar(product,sales,color='green')
plt.xlabel('Products')
plt.ylabel('Sale Quantity')
plt.title('Product Sales Comparision')
plt.show()
```

```python
plt.barh(product,sales,color='red')
plt.xlabel('Products')
plt.ylabel('Sale Quantity')
plt.title('Product Sales Comparision')
plt.show()
```

```python
colors=['red','yellow','green','blue']
exp=[0.01,0.01,0.1,0.01]
plt.pie(sales,labels=product,autopct="%0.2f%%",colors=colors,explode=exp,radius=1.2)
plt.title('Pie Chart')
plt.show()
```

```python
students = ['Ali', 'Ahmad']

math = [7, 4]
physics = [8, 6]
comp = [6, 9]
total = [21, 19]

x = np.arange(len(students))
width = 0.2

plt.bar(x - 1.5*width, math, width, label='Math')
plt.bar(x - 0.5*width, physics, width, label='Physics')
plt.bar(x + 0.5*width, comp, width, label='Computer')
plt.bar(x + 1.5*width, total, width, label='Total')

plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Students Marks Comparison')
plt.savefig('marks_com_jahanzeb_bar_chart.jpg')
plt.xticks(x, students)
plt.legend()
plt.show()
```

```python


students = ['Ali', 'Ahmad']

math = [7, 4]
physics = [8, 6]
comp = [6, 9]
total = [21, 19]

x = np.arange(len(students))
width = 0.2

# Create 2 graphs in one row
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# Graph 1: Grouped Bar Chart
b1 = ax[0].bar(x - 1.5*width, math, width, label='Math')
b2 = ax[0].bar(x - 0.5*width, physics, width, label='Physics')
b3 = ax[0].bar(x + 0.5*width, comp, width, label='Computer')
b4 = ax[0].bar(x + 1.5*width, total, width, label='Total')

# Show values on grouped bars
ax[0].bar_label(b1, padding=2)
ax[0].bar_label(b2, padding=2)
ax[0].bar_label(b3, padding=2)
ax[0].bar_label(b4, padding=2)

ax[0].set_xlabel('Students')
ax[0].set_ylabel('Marks')
ax[0].set_title('Chart By Jahanzeb Mughal')
ax[0].set_xticks(x)
ax[0].set_xticklabels(students)
ax[0].legend()

# Graph 2: Stacked Bar Chart
s1 = ax[1].bar(x, math, label='Math')
s2 = ax[1].bar(x, physics, bottom=math, label='Physics')

bottom_comp = np.array(math) + np.array(physics)
s3 = ax[1].bar(x, comp, bottom=bottom_comp, label='Computer')

# Show values inside stacked bars
ax[1].bar_label(s1, label_type='center')
ax[1].bar_label(s2, label_type='center')
ax[1].bar_label(s3, label_type='center')

ax[1].set_xlabel('Students')
ax[1].set_ylabel('Marks')
ax[1].set_title('Chart By Jahanzeb Mughal')
ax[1].set_xticks(x)
ax[1].set_xticklabels(students)
ax[1].legend()

# Main title
fig.suptitle('Students Marks Comparison', fontsize=16)

plt.tight_layout()
plt.savefig('both_bar_charts.jpg', dpi=300, bbox_inches='tight')
plt.show()
```

# Building EDA Report Using Subplot

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 6))

ax[0, 0].plot([1, 2, 3], [10, 20, 30])
ax[0, 0].set_title("Line Plot")

ax[0, 1].bar([1, 2, 3], [10, 20, 15])
ax[0, 1].set_title("Bar Plot")

ax[1, 0].scatter([1, 2, 3, 4], [5, 15, 10, 24])
ax[1, 0].set_title("Scatter Plot")

ax[1, 1].hist([10, 20, 20, 30, 30, 30])
ax[1, 1].set_title("Histogram")

plt.tight_layout()
plt.show()
```

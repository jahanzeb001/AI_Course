# Exploratory Data Analysis (Lect 04)

```python
# imports and setup
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('darkgrid')
```

## Fundamentals of Seaborns

```python
df=sns.load_dataset('tips')
```

```python
df.head()
```

### Line Plot

```python
plt.figure(figsize=(20,4))
sns.lineplot(x='total_bill',y='tip',data=df)
plt.title('Total Bill vs Tip')
plt.savefig('lineplot.png')
plt.show()
```

### Bar Plot

```python
plt.figure(figsize=(20,4))
sns.barplot(x='day',y='total_bill',data=df)
plt.title("Day Wise Total Bill")
plt.savefig('barplot.png')
plt.show()
```

### Count plot 

```python
sns.countplot(x='day',data=df)
plt.title("Count Days")
plt.savefig('countplot.png')
plt.show()
```

### Box Plot

```python
plt.figure(figsize=(20,4))
sns.boxplot(x='day',y='total_bill',data=df)
plt.title("Day Vs Total Bill")
plt.savefig('boxplot.png')
plt.show()
```

### Violin Plot

```python
plt.figure(figsize=(20,4))
sns.violinplot(x='day',y='total_bill',data=df)
plt.title("Day Vs Total Bill")
plt.savefig('violin.png')
plt.show()
```

### Swarm plot

```python
plt.figure(figsize=(20,4))
sns.swarmplot(x='day',y='total_bill',data=df)
plt.title("Day Vs Total Bill")
plt.savefig('violin.png')
plt.show()
```

### Strip plot

```python
plt.figure(figsize=(20,4))
sns.stripplot(x='day',y='total_bill',data=df,jitter=False)
plt.title("Day Vs Total Bill")
plt.savefig('violin.png')
plt.show()
```

### Histograme

```python
plt.figure(figsize=(20,4))
sns.histplot(df['total_bill'],bins=30, kde=True)
plt.title("Frequency Count of Total bill")
plt.savefig('histplot.png')
plt.show()
```

### KDE PLOT

```python
plt.figure(figsize=(20,4))
sns.kdeplot(df['total_bill'])
plt.title("Frequency Distribution of Total bill")
plt.savefig('kdeplot.png')
plt.show()
```

### Heat Map

```python
corr=df.corr(numeric_only=True)
```

```python
sns.heatmap(corr,annot=True)
```

```python
plt.figure(figsize=(20,4))
sns.heatmap(corr,annot=True)
plt.title("Heat Map of Correlation Matrix")
plt.savefig('heatmap.png')
plt.show()
```

### Pair Plot

```python
sns.pairplot(df)
plt.title('Pair Plot')
```
